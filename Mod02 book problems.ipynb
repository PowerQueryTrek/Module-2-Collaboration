## 4.1 Guess the number

```python
# 4.1 Choose secret and guess; print too low / too high / just right

import random

secret = random.randint(1, 10)  # random picks a number
guess = int(input("guess a whole number between 1 and 10 inclusive, e.g., 5:"))

if guess == secret:
    print('just right')
elif guess < secret:
    print ('too low')
else:
    print ('too high')
```

## 4.2 True or False example

```python
# 4.2 Boolean logic example

small = True
green = False

if small:
    if not green:
        print('cherry')
    else:
        print ('pea')
else:
    if not green:
        print('pumpkin')
    else:
        print ('watermelon')
```

## 6.1 For loop to print set elements

```python
numbers = [3, 2, 1, 0]
for number in numbers:
    print(number)
```

## 6.2 Guess with a while loop

```python
guess_me = 7
number = 1

while True:
    if number < guess_me:
        print("too low")
    elif number == guess_me:
        print("found it!")
        break
    else:
        print("oops")
        break
    number += 1       
```

## 6.3 Guess with a for loop

```python
guess_me = 5

for number in range(10):
    if number < guess_me:
        print("too low")
    elif number == guess_me:
        print("found it!")
        break
    else:
        print("oops")
        break
    number += 1      
```
